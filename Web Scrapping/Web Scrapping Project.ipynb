```python
import re
import requests
import time
import pandas as pd
from bs4 import BeautifulSoup
from urllib.parse import urljoin

base_url = "https://books.toscrape.com/catalogue/category/books_1/"
rating_map = {"One":1,
              "Two":2,
              "Three":3,
              "Four":4,
              "Five":5}

book_list = []

for page_no in range(1,51):
  if page_no == 1:
    url = base_url + "index.html"
  else:
    url = base_url + f"page-{page_no}.html"

  response = requests.get(url)
  soup = BeautifulSoup(response.content,"html.parser")
  book_card = soup.find_all("article",class_="product_pod")

  print("Scrapping page no :",page_no)

  for books in book_card:
    Title = books.find("h3").find("a")["title"]
    stock = books.find("p",class_="instock availability").text.strip()
    rating = books.find("p",class_="star-rating")["class"][1]
    star_no = rating_map[rating]
    star = "\u2b50" * star_no
    price = books.find("p",class_="price_color").text.strip()

    description_url = books.find("h3").find("a")["href"]
    final_url = urljoin(base_url,description_url)

    book_response = requests.get(final_url)
    book_soup = BeautifulSoup(book_response.content,"html.parser")
    book_info = book_soup.find("article",class_="product_page")

    book_category = book_soup.find("ul",class_="breadcrumb").find_all("li")
    category = book_category[2].text.strip()

    desc_div = book_soup.find("div", id="product_description")
    if desc_div:
      description = desc_div.find_next("p").text.strip()
    else:
      description = "No description available"

    product_code = book_soup.find("th", string="UPC").find_next("td").text.strip()
    product_type = book_soup.find("th", string="Product Type").find_next("td").text.strip()
    reviews = book_soup.find("th", string="Number of reviews").find_next("td").text.strip()
    tax = book_soup.find("th", string="Tax").find_next("td").text.strip()

    stock_text = book_soup.find("th", string="Availability").find_next("td").text.strip()
    match = re.search(r"\((\d+)\s+available\)", stock_text)
    if match:
      stock_quantity = int(match.group(1))
    else:
      stock_quantity = 0

    book_list.append({
    "Product Code": product_code,
    "Title": Title,
    "Price":price,
    "Category":category,
    "Stock": stock,
    "Stock Quantity": stock_quantity,
    "Tax": tax,
    "Rating":star,
    "No of Reviews": reviews,
    "Description": description

  })

    #print(Title)
    #print(category)
    #print(star)
    #print("-" * 20)
    time.sleep(1)

df = pd.DataFrame(book_list)
df.to_excel("Book_List.xlsx",index="true")

```
